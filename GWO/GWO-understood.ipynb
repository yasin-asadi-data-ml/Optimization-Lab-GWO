# GWO

```python
import numpy as np  # NumPy provides fast vectorized math and array operations

# -----------------------------
# 🧠 Objective (fitness) function
# -----------------------------
# This is the mathematical environment — the "hunting field" — where the wolves search for the prey.
# You can replace this with any function whose minimum (or maximum) you want to find.
def objective_function(x):
    # Example: simple convex function f(x) = sum(x^2)
    # Intuitively: the closer x is to zero, the smaller the value — so wolves are "attracted" toward origin.
    return np.sum(x ** 2)


# -----------------------------
# 🐺 Grey Wolf Optimizer (GWO)
# -----------------------------
def GWO(obj_func, dim, search_agents_no, max_iter, lower_bound, upper_bound):
    """
    Philosophical core of the GWO:
    Wolves represent candidate solutions.
    They form a social hierarchy:
        α (alpha): best (leader)
        β (beta): second best (advisor)
        δ (delta): third best (hunter)
    Others are ω (omega): followers
    The pack hunts collectively toward the prey (optimal solution).
    """

    # Initialize positions of wolves (search agents)
    # This creates a matrix of shape [number_of_wolves, dimension_of_problem]
    # Each element is randomly chosen between lower_bound and upper_bound
    positions = np.random.uniform(lower_bound, upper_bound, (search_agents_no, dim))

    # Initialize alpha, beta, delta — the top three best solutions
    # They start as zero vectors in the same dimension as the problem
    alpha_pos = np.zeros(dim)
    beta_pos = np.zeros(dim)
    delta_pos = np.zeros(dim)

    # Initialize their fitness values to +infinity (since we seek minimum)
    alpha_score = float("inf")
    beta_score = float("inf")
    delta_score = float("inf")

    # Iterative process — wolves adjust their positions step by step
    for iteration in range(max_iter):

        # Evaluate fitness of each wolf (i.e., how close it is to the prey)
        for i in range(search_agents_no):

            # Enforce boundaries (wolves can’t go beyond the hunting field)
            positions[i] = np.clip(positions[i], lower_bound, upper_bound)

            # Calculate fitness
            fitness = obj_func(positions[i])

            # Update hierarchy based on fitness
            if fitness < alpha_score:
                # Found a new best (alpha)
                alpha_score = fitness
                alpha_pos = positions[i].copy()
            elif fitness < beta_score:
                # Found second best (beta)
                beta_score = fitness
                beta_pos = positions[i].copy()
            elif fitness < delta_score:
                # Found third best (delta)
                delta_score = fitness
                delta_pos = positions[i].copy()

        # The coefficient 'a' decreases linearly from 2 to 0
        # Symbolically, it represents the balance between exploration and exploitation.
        # High 'a' → more randomness (exploring new places)
        # Low 'a' → more exploitation (fine-tuning near alpha)
        a = 2 - iteration * (2 / max_iter)

        # Now, each wolf updates its position based on α, β, and δ
        for i in range(search_agents_no):
            for j in range(dim):

                # Compute random coefficients r1 and r2 — they simulate uncertainty in the wolf’s thinking.
                r1 = np.random.rand()
                r2 = np.random.rand()

                # Compute coefficient A — determines *how strongly* the wolf is pulled toward the prey
                # A = 2*a*r1 - a  ⇒ changes between [-a, +a]
                A1 = 2 * a * r1 - a
                A2 = 2 * a * r1 - a
                A3 = 2 * a * r1 - a

                # Compute coefficient C — adds a random distortion around the leader’s position
                # It multiplies the leader’s position to introduce “hunting noise”
                C1 = 2 * r2
                C2 = 2 * r2
                C3 = 2 * r2

                # Compute distances to alpha, beta, delta (prey estimations)
                D_alpha = abs(C1 * alpha_pos[j] - positions[i][j])
                D_beta = abs(C2 * beta_pos[j] - positions[i][j])
                D_delta = abs(C3 * delta_pos[j] - positions[i][j])

                # Compute candidate positions from each of the three leaders
                X1 = alpha_pos[j] - A1 * D_alpha
                X2 = beta_pos[j] - A2 * D_beta
                X3 = delta_pos[j] - A3 * D_delta

                # The wolf’s new position is the mean of these three — reflecting social cooperation.
                positions[i][j] = (X1 + X2 + X3) / 3

        # Optional: print progress
        print(f"Iteration {iteration+1}/{max_iter} | Alpha Fitness: {alpha_score:.6f}")

    # Return best solution and its score
    return alpha_pos, alpha_score


# -----------------------------
# ⚙️ Run GWO
# -----------------------------
# Define problem settings
dim = 5                 # Problem has 5 variables (dimensions)
search_agents_no = 10   # 10 wolves (agents)
max_iter = 30           # 30 iterations (time steps)
lower_bound = -10       # Minimum value each variable can take
upper_bound = 10        # Maximum value each variable can take

# Run the optimizer
best_position, best_score = GWO(objective_function, dim, search_agents_no, max_iter, lower_bound, upper_bound)

# Print final result
print("\n✅ Best Solution Found:")
print("Position (variables):", best_position)
print("Fitness (objective value):", best_score)
```
